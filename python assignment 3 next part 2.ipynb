```python
import math

try:
    a = float(input("Enter coefficient a: "))
    b = float(input("Enter coefficient b: "))
    c = float(input("Enter coefficient c: "))

    if a == 0:
        print("Not a quadratic equation")
    else:
        d = b**2 - 4*a*c

        if d < 0:
            print("No real roots")
        else:
            root1 = (-b + math.sqrt(d)) / (2*a)
            root2 = (-b - math.sqrt(d)) / (2*a)

            print("Root 1 =", root1)
            print("Root 2 =", root2)

except ValueError as e:
    print("ValueError:", e)
```
