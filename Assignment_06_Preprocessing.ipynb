```python
import pandas as pd
import numpy as np
```

```python
# Sample data with missing values, a duplicate row and an irrelevant column
data = {
    "Name":    ["Amit", "Bhavna", "Chetan", "Diya", "Esha", "Bhavna", "Farhan", "Gauri"],
    "Marks":   [75, 82, np.nan, 90, 72, 82, 85, np.nan],
    "City":    ["Pune", "Mumbai", "Pune", "Nashik", None, "Mumbai", "Pune", "Mumbai"],
    "RollNo":  [101, 102, 103, 104, 105, 102, 106, 107],
    "Remarks": ["-", "-", "-", "-", "-", "-", "-", "-"],
}
df = pd.DataFrame(data)
print(df)
```

```python
# Explore the data
print(df.head(3))
print()
df.info()
print()
print(df.describe())
```

```python
# Missing values
print(df.isnull().sum())
```

```python
# dropna()
print(df.dropna())
```

```python
# fillna(): mean for marks, mode for city
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["City"] = df["City"].fillna(df["City"].mode()[0])
print(df)
```

```python
# Duplicates
print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
print(df)
```

```python
# Drop irrelevant column
df = df.drop(columns=["Remarks"])
print(df)
```

```python
# Filtering with loc[]
print("Students with Marks > 80:")
print(df.loc[df["Marks"] > 80])
print()
print("Name and City of students from Pune:")
print(df.loc[df["City"] == "Pune", ["Name", "City"]])
```
