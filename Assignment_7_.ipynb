```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the CSV dataset (file name can also be taken from the user with input())
fname = "sample_data_assignment_07.csv"
# fname = input("Enter CSV file name: ")
df = pd.read_csv(fname)
print("Dataset:")
print(df)
```

```python
# Line plot
plt.figure(figsize=(7, 4.5))
plt.plot(df["Student"], df["Python"], marker="o", label="Python")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Bar plot
plt.figure(figsize=(7, 4.5))
plt.bar(df["Student"], df["Maths"], label="Maths")
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.legend()
plt.show()
```

```python
# Scatter plot
plt.figure(figsize=(7, 4.5))
plt.scatter(df["Maths"], df["Python"], s=70, label="Students")
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Correlation matrix and heatmap
correlation = df[["Maths", "Python", "Attendance"]].corr()
print("Correlation Matrix:")
print(correlation)

plt.figure(figsize=(6.5, 5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Challenge: histogram of Python marks
plt.figure(figsize=(7, 4.5))
plt.hist(df["Python"], bins=5, edgecolor="black", label="Python")
plt.title("Histogram of Python Marks")
plt.xlabel("Python Marks")
plt.ylabel("Number of Students")
plt.legend()
plt.show()
```

```python
# Challenge: box plot of Maths and Python marks
plt.figure(figsize=(7, 4.5))
sns.boxplot(data=df[["Maths", "Python"]])
plt.title("Box Plot of Maths and Python Marks")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()
```

```python
# Challenge: pie chart of students per city
counts = df["City"].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(counts, labels=counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Students per City")
plt.legend(counts.index, title="City")
plt.show()
```

```python
# Challenge: averages
print("Average Maths marks     :", round(df["Maths"].mean(), 2))
print("Average Python marks    :", round(df["Python"].mean(), 2))
print("Average Attendance      :", round(df["Attendance"].mean(), 2))
```
